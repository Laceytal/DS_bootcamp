#### 1. Create a dataframe called views with two columns: datetime and user by reading feed-views.log.

```python
import pandas as pd
views = pd.read_csv('../data/feed-views.log',
                    sep='\t',
                    names=['datetime', 'user'],
                    index_col='datetime',
                    engine='python')
views.index = pd.to_datetime(views.index)

views['year']= views.index.year
views['month']= views.index.month
views['day']= views.index.day
views['hour']= views.index.hour
views['minute']= views.index.minute
views['second']= views.index.second

views.head()
```

#### 2. Create the new column daytime.

```python
dayTimes = ['night','early morning','morning','afternoon', 'early evening','evening']
dayTimings= [0,4,7,11,17,20,24]
views['daytime']= pd.cut(
    views['hour'],
    bins=dayTimings,
    labels=dayTimes,
    right=False
)
views= views.set_index('user')
views[5:15]
```

#### 3. Calculate the number of elements in your dataframe.

```python
views['daytime'].value_counts() 
```

#### 4. Sort the values in your dataframe by hour, minute, and second in ascending order simultaneously, not one by one.

```python
views.sort_values(['hour','minute','second'],inplace=True)
views.head()
```

#### 5. Calculate the minimum and maximum for the hours and the mode for the daytime categories.

```python
night= views[views['daytime']=='night']
max_night= night[['hour']].max()
print("Max for night:\n",max_night)


print("="*80)

morning= views[views['daytime']=='morning']
min_morning= morning[['hour']].min()
print("\nMin for morning:\n",min_morning,'\n')

print("="*80)

night_max_hour_visit= night[night['hour']==max_night['hour']]
print('\n',night_max_hour_visit[0:1:],'\n')

print("="*80)

morning_min_hour_visit= morning[morning['hour']==min_morning['hour']]
print('\n',morning_min_hour_visit[0:1],'\n')

print("="*80,'\n')

hour_mode=views["hour"].mode()
print('Hour mode:\n',hour_mode,'\n')

print("="*80,'\n')

daytime_mode= views["daytime"].mode()
print('Daytime mode:\n',daytime_mode)
```

#### 6. Show the three earliest and latest hours of the day and their corresponding usernames using nsmallest() and nlargest().

```python
earliest = views.nsmallest(3, 'hour')
earliest
```

```python
latest= views.nlargest(3, 'hour')
latest
```
