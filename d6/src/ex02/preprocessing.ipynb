#### 1. Download and read the CSV file, making ID the index column.

```python
import pandas as pd
df = pd.read_csv('../data/auto.csv',
                 sep=',',
                 engine='python',
                 index_col='ID',
                 names=['ID','CarNumber','Make_n_model','Refund','Fines','History'],
                 skiprows=1)
df
```

#### 2. Count the number of observations using the method count().


```python
number_of_records = df.count()
number_of_records
```

#### 3. Drop the duplicates, taking into account only the following columns: CarNumber, Make_n_Model, and Fines.

```python
df.drop_duplicates(subset=['CarNumber','Make_n_model','Fines'], keep='last', inplace=True)
number_of_records=df.count()
number_of_records
```

#### 4. Work with missing values.

```python
df.isna().sum()
```

```python
df.dropna(axis=1,thresh=len(df)-500,inplace=True)
df.isna().sum()
```

```python
df['Refund'] = df['Refund'].ffill()
df.isna().sum()
```

```python
df['Fines'] = df['Fines'].fillna(df['Fines'].mean())

df.isna().sum()
```

#### 5. Split and parse the make and model.


```python
df['Make']= df['Make_n_model'].apply(lambda x : x.split(' ')[0])
df['Model']= df['Make_n_model'].apply(lambda x : ''.join(x.split(' ')[1:]))
df.drop(columns=['Make_n_model'],inplace=True)
```

```python
df.to_json('../data/auto.json',orient='records')
```
