#### 1. Read:

```python
import pandas as pd
df = pd.read_csv('../data/feed-views.log',
                 sep='\t',
                 names=['datatime', 'user'], 
                 skiprows=[2,3],skipfooter=2,
                 engine='python',
                 index_col='datatime')
```

#### 2. Rename:

```python
df.index.name = 'data_time'
```

#### 3. To csv:

```python
df.to_csv('../data/feed-views-semicolon.log', sep=';')
```
